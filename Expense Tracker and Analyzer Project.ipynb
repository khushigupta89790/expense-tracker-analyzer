# Expense Tracker and Analyzer

```python
import csv
```

```python
def load_expenses():
    expenses = []

    try:
        with open('expenses.csv', 'r', newline='') as file:
            reader = csv.DictReader(file)

            for row in reader:
                row['amount'] = float(row['amount'])
                expenses.append(row)

    except FileNotFoundError:
        pass

    return expenses
```

```python
def save_expenses(expenses):
    with open('expenses.csv', 'w', newline='') as file:
        fieldnames = ['amount', 'category', 'description']

        writer = csv.DictWriter(file, fieldnames=fieldnames)

        writer.writeheader()
        writer.writerows(expenses)
```

```python
def add_expense():
    
    x = float(input('Add Amount: '))
    y = input('Category: ')
    z = input('Description of expense: ')
    expense = {'amount':x,'category':y,'description':z}
    expenses.append(expense)
    save_expenses(expenses)
```

```python
def view_expenses(expenses):
    for i,money in enumerate(expenses,start=1):
        amount = money['amount']
        category = money['category']
        description = money['description']
        print('Expense',i)
        print('Amount:',amount)
        print('Category:',category)
        print('Description:',description)
        print('\n')
```

```python
def display_menu():
    print('----------MENU----------')
    print('1. Add Expense')
    print('2. View Expense')
    print('3. Analyze')
    print('4. Exit')
    print('\n')
```

```python
def analyze_expenses(expenses):
    total = 0 
    highest = 0 
    category_totals = {}

    for expense in expenses:
        total += expense['amount']

        if expense['amount'] > highest:
            highest = expense['amount']

        category = expense['category']

        if category in category_totals:
            category_totals[category] += expense['amount']
        else:
            category_totals[category] = expense['amount']

    
    highest_category = None
    highest_amount = 0 

    for category, amount in category_totals.items():
        if amount > highest_amount:
            highest_amount = amount
            highest_category = category

    print('Total Expenses:', total)
    print('Highest Expense:', highest)
    print(f'Highest spending category was {highest_category} and the amount spent was {highest_amount}')
    print('\n')
```

```python
expenses = load_expenses()
```

```python
choice = 0

while choice != 4:

    display_menu()

    choice = int(input('Enter a choice(1-4): '))

    if choice == 1:
        add_expense()

    elif choice == 2:
        view_expenses(expenses)

    elif choice == 3:
        analyze_expenses(expenses)

    elif choice == 4:
        print('Exiting...')

    else:
        print('Invalid Choice! Please enter a number between 1 and 4')       
```
